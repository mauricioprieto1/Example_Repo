#Problema de chocolates

```python
from scipy.stats import t
import pandas as pd
d = {'peso':[96.4,97.67,98.48,97.67,100.11,95.29,99.88,98.8,100.53,99.41,97.64,101.11,93.43,96.99,97.92]}
datos = pd.DataFrame(d)
#la H1 es promedio de chocolates diferentes de 100g
x = datos['peso'].mean()
std = datos['peso'].std()
gradLib = datos['peso'].count()-1
errStd = std/(datos['peso'].count()**0.5)
pvalue = 2*t.cdf(x,gradLib,100,errStd)
print('promedio',x, 'std',std, 'pvalue',pvalue)
#Cambio de hipotesis, H1: Chocolates menor peso de 100
pvalue = t.cdf(x,gradLib,100,errStd)
print('promedio',x, 'std',std, 'pvalue',pvalue)
```

#Problema panadero y panadera

```python
from scipy import stats
import pandas as pd
papa = {'peso':[499, 500,510,495,515,503,490,511]}
mama = {'peso':[512,530,498,540,521,528,505,523]}
dataPapa = pd.DataFrame(papa)
dataMama = pd.DataFrame(mama)
print('el promedio papa: ',dataPapa['peso'].mean())
print('el promedio mama: ',dataMama['peso'].mean())
#H1: peso galletas panadera es mayor a peso galletas panadero
pvalue = stats.ttest_ind(dataMama['peso'],dataPapa['peso'],equal_var= False, alternative='greater')
print(pvalue)
```

```python
#Comparacion informal sobre las medias
#Quiero saber cuales medias son diferentes
#Priemro un box plot
import pandas as pd
from statsmodels.formula.api import ols
import matplotlib.pyplot as plt
import scipy.stats as stats
data = pd.read_csv('libido.csv')
dataBox = data.groupby('dose')
dataBox.boxplot(column = 'libido',subplots=False)
plt.show()
#-------------------------------------
#Voy a utilizar intervalos de confianza
placebo = data[data['dose']=='placebo']['libido']
low = data[data['dose']=='low']['libido']
high = data[data['dose']=='high']['libido']
confInt = 0.5
dF = len(placebo)-1
#intervalo de confianza
confIntPlacebo = stats.t.interval(confInt,dF,placebo.mean(),placebo.std())
confIntLow = stats.t.interval(confInt,dF,low.mean(),low.std())
confIntHigh = stats.t.interval(confInt,dF,high.mean(),high.std())
print("Intervalo de confianza 50% placebo: ", confIntPlacebo)
print("Intervalo de confianza 50% low: ", confIntLow)
print("Intervalo de confianza 50% high: ", confIntHigh)
```

```python
#Ejemplo bonferroni libido
import statsmodels.stats.multicomp as mc
import pandas as pd
import scipy.stats as stats
data = pd.read_csv('libido.csv')
comp = mc.MultiComparison(data['libido'],data['dose'])
tbl,al,a2 = comp.allpairtest(stats.ttest_ind,method ='bonf')
print(tbl)
```

#Ejemplo detergente anova factor 2

```python
import pandas as pd
from statsmodels.formula.api import ols
import statsmodels.api as sm
import statsmodels.stats.multicomp as mc
df = pd.read_csv('detergente (1).csv')
model = ols('limpieza ~ C(detergente)+C(temperatura)+C(detergente):C(temperatura)',data=df).fit()
print(sm.stats.anova_lm(model,typ=2))

#Pruebas Tukey Detergentes (Factor A)--------------------------------|
comp = mc.MultiComparison(df['limpieza'],df['detergente'])
out = comp.tukeyhsd(0.05)
print(out)
#Pruebas Tukey Temperaturas (Factor B)---------------------------------|
comp = mc.MultiComparison(df['limpieza'],df['temperatura'])
out = comp.tukeyhsd(0.05)
print(out)
#Pruebas Tukey Temperaturas (factor AB)-----------------------------|
comp = mc.MultiComparison(df['limpieza'],df['temperatura']+df['detergente'])
out = comp.tukeyhsd(0.05)
print(out)
```

```python
import pandas as pd
from statsmodels.formula.api import ols
import statsmodels.api as sm
import statsmodels.stats.multicomp as mc
df = pd.read_csv('life.csv')
model = ols('satisfaccion ~ C(sexo)+C(edad)+C(sexo):C(edad)',data=df).fit()
print(sm.stats.anova_lm(model,typ=2))

#Pruebas Tukey Detergentes (Factor A)-------------------------------|
comp = mc.MultiComparison(df['satisfaccion'],df['edad'])
out = comp.tukeyhsd(0.05)
print(out)
#Pruebas Tukey Temperaturas (Factor B)------------------------------|
comp = mc.MultiComparison(df['satisfaccion'],df['edad'])
out = comp.tukeyhsd(0.05)
print(out)
#Pruebas Tukey Temperaturas (factor AB)-----------------------------|
comp = mc.MultiComparison(df['satisfaccion'],df['sexo']+df['edad'])
out = comp.tukeyhsd(0.05)
print(out)
```

```python
#Pruebas de hipotesis
import pandas as pd
from scipy import stats

df = pd.read_csv('theatre-1.csv')

culture_male = df[df['Sex'] == 0]['Culture']
culture_female = df[df['Sex'] == 1]['Culture']

promedio_culture_male = culture_male.mean()
promedio_culture_female = culture_female.mean()

t_stat_a, p_value_a = stats.ttest_ind(culture_male, culture_female, equal_var=False)

print("(a) Promedio de gasto en actividades culturales:")
print(f"Hombres: {promedio_culture_male:}, Mujeres: {promedio_culture_female:}")
print(f" {p_value_a:}")

theatre_male = df[df['Sex'] == 0]['Theatre']
theatre_female = df[df['Sex'] == 1]['Theatre']

mean_theatre_male = theatre_male.mean()
mean_theatre_female = theatre_female.mean()

t_stat_b, p_value_b = stats.ttest_ind(theatre_male, theatre_female, equal_var=False)

print("(b) Promedio de gasto en teatro este año:")
print(f"Hombres: {mean_theatre_male:.2f}, Mujeres: {mean_theatre_female:}")
print(f"{p_value_b:}")

theatre_current = df['Theatre']
theatre_last_year = df['Theatre_ly']

t_stat_c, p_value_c = stats.ttest_rel(theatre_current, theatre_last_year)

print("(c) Prueba de hipótesis entre gasto en teatro este año vs. año pasado (global):")
print(f"{p_value_c:}")

t_stat_d_male, p_value_d_male = stats.ttest_rel(theatre_male, df[df['Sex'] == 0]['Theatre_ly'])

t_stat_d_female, p_value_d_female = stats.ttest_rel(theatre_female, df[df['Sex'] == 1]['Theatre_ly'])

print("(d) Prueba de hipótesis entre gasto este año vs. año pasado (hombres):")
print(f"{p_value_d_male:}")

print("Prueba de hipótesis entre gasto este año vs. año pasado (mujeres):")
print(f"{p_value_d_female:}")
```

```python
#Actividad Anova
import pandas as pd
from statsmodels.formula.api import ols
import statsmodels.api as sm
import statsmodels.stats.multicomp as mc
import scipy.stats as stats
import matplotlib.pyplot as plt

df = pd.read_csv('javalina.csv')

print(df.head())

modelo = ols('distancia ~ C(metodo)', data=df).fit()
anova_table = sm.stats.anova_lm(modelo, typ=2)
print("Tabla ANOVA:")
print(anova_table)

comparaciones = mc.MultiComparison(df['distancia'], df['metodo'])
bonferroni_result = comparaciones.allpairtest(stats.ttest_ind, method='bonf')
print("\nResultados de la prueba de Bonferroni:")
print(bonferroni_result[0])

df.boxplot(column='distancia', by='metodo', grid=False)
plt.title('Distribución de las distancias por método de lanzamiento')
plt.suptitle('')
plt.ylabel('Distancia (metros)')
plt.show()
```

# Regresion Lineal

```python
#Ejemplo de terapia
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
import matplotlib.pyplot as plt

datos = pd.read_csv('terapia.csv')
model=ols('tiempoRecuperacion ~ tiempoTerapia', data=datos).fit()
prediccion = model.predict(datos['tiempoTerapia'])#y=a+bx
resultados=model.summary()
print(resultados)
```

#Graficas

```python
#graficas
x=datos['tiempoTerapia']
y=model.fittedvalues #los valorwes calculados
yo = datos['tiempoRecuperacion']
plt.plot(x,y,color='red',label='Modelo')
plt.scatter(x,yo,color='k',label='Datos')
plt.legend(['Modelo','Datos'])
plt.show()
residuales = model.resid
plt.scatter(x,residuales,color='k') #Grafica residuales
plt.show()
plt.hist(residuales,color='k') #histograma de los residuales
plt.show()
```

# Ejemplo Naive Bayes (golf)

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn import metrics
from sklearn.metrics import confusion_matrix

datos = pd.read_csv('NaiveBayes_ejemplo2.csv')
X = datos.iloc[:,:4].values

Y = datos['PlayGolf'].values
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.5, random_state=1)

mnb = MultinomialNB()
mnb.fit(X_train,y_train)

y_pred = mnb.predict(X_test)
resultado = metrics.accuracy_score(y_test,y_pred)*100
print('Modelo multinomial, Naive-Bayes exactitud(%):',resultado)

#Lo de confusion feo
confusion = confusion_matrix(y_test,y_pred)
print(confusion)

#Lo de confusion bonito
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(confusion,annot=True,cmap='Blues')
plt.show()
#outlook, temperature, humidity, windy, playgolf
prediccion = mnb.predict([[0,0,0,0]])
print('Juego golf:', prediccion)
```

#Clasificacion NaiveBayes_cancer de mama

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn import metrics
from sklearn.metrics import confusion_matrix

datos = pd.read_csv('breast_cancer.csv')
X = datos.iloc[:,2:].values

Y = datos['diagnosis'].values
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=1)

gnb = GaussianNB()
gnb.fit(X_train,y_train)

y_pred = gnb.predict(X_test)
resultado = metrics.accuracy_score(y_test,y_pred)*100
print('Modelo Gaussiano, Naive-Bayes exactitud(%):',resultado)

#Lo de confusion feo
confusion = confusion_matrix(y_test,y_pred)
print(confusion)

#Lo de confusion bonito
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(confusion,annot=True,cmap='Blues')
plt.show()
#outlook, temperature, humidity, windy, playgolf
prediccion = gnb.predict([[0,0,0,0]])
print('Juego golf:', prediccion)
```

Arbol de desiciones (ejemplo golf)

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
import seaborn as sns
from sklearn import metrics

data = pd.read_csv('NaiveBayes_ejemplo2.csv')
X = data.iloc[:,:4].values #Columnas para la prediccion (nodos)
Y = data['PlayGolf'].values #Valor de la variable predicha
#dividir los datos en train y test
X_train, X_test, Y_train, Y_test=train_test_split(X, Y, test_size=0.5, random_state=1)
tree = DecisionTreeClassifier()#define la variable
tree.fit(X_train,Y_train) #Entrena el arbol
Y_pred = tree.predict(X_test)
print('Exactitud del arbol: ', metrics.accuracy_score(Y_test,Y_pred))
confusion = confusion_matrix(Y_test,Y_pred)
print(confusion)
sns.heatmap(confusion,annot=True,cmap='Blues')
plt.show()
#
```

#Hacer png del arbol

```python
from sklearn.tree import export_graphviz
from six import StringIO
from IPython.display import Image
import pydotplus
dot_data = StringIO()
export_graphviz(tree,out_file=dot_data,filled=True,rounded=True,special_characters=True, feature_names=['Outlook', 'Temperatura', 'Humidity', 'Windy'],class_names=['NoGolf','YesGolf'])
graph = pydotplus.graph_from_dot_data(dot_data.getvalue())
graph.write_png('golf.png')
Image(graph.create_png())
#
```

```python
#Cual variable es mas importante para jugar golf
plt.barh(data.iloc[:,:4].columns.values,tree.feature_importances_)
plt.show()
#
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
import seaborn as sns
from sklearn import metrics

data = pd.read_csv('breast_cancer.csv')
columnas = data.columns.drop(['id', 'diagnosis'])
X = data[columnas.values].values#Columnas para la prediccion (nodos)
Y = data['diagnosis'].values #Valor de la variable predicha
#dividir los datos en train y test
X_train, X_test, Y_train, Y_test=train_test_split(X, Y, test_size=0.3, random_state=1)
tree = DecisionTreeClassifier(max_depth=3)#define la variable
tree.fit(X_train,Y_train) #Entrena el arbol
Y_pred = tree.predict(X_test)
print('Exactitud del arbol: ', metrics.accuracy_score(Y_test,Y_pred))
confusion = confusion_matrix(Y_test,Y_pred)
print(confusion)
sns.heatmap(confusion,annot=True,cmap='Blues')
plt.show()
#
```

#Ejemplo Ventas en linea
nos interesan las columnas codigo postal,categoria sales, cantidad, descuento y trataremos de predicir si hay profit

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
import seaborn as sns
from sklearn import metrics
from sklearn.preprocessing import LabelEncoder

datos = pd.read_csv('sales.csv')
label_encoder = LabelEncoder()
#esta linea transforma la columa 'Categoria' de variable categorica
#a variable numerica
datos['Category'] = label_encoder.fit_transform(datos['Category'])
datos[datos['Profit']>0]['Profit'].describe()
#agregar una columa de ganancia yes/no
#yes sera una ganancia de mayor de 100
#no sera una ganancia menor de 100
def profBool(num):
  if num>100:
    return 1
  else:
    return 0
  datos['ProfitBool'] = datos['Profit'].apply(profBool)

columnas = ['PostalCode', 'Category','Sales','Quantity','Discount']
X = datos[columnas].values
Y = datos['ProfitBool'].values

X_train, X_test, Y_train, Y_test=train_test_split(X, Y, test_size=0.3, random_state=1)
tree = DecisionTreeClassifier(max_depth=3)#define la variable
tree.fit(X_train,Y_train) #Entrena el arbol
Y_pred = tree.predict(X_test)
print('Exactitud del arbol: ', metrics.accuracy_score(Y_test,Y_pred))
confusion = confusion_matrix(Y_test,Y_pred)
print(confusion)
sns.heatmap(confusion,annot=True,cmap='Blues')
plt.show()
```

```python
plt.barh(columnas, tree.feature_importances_)
plt.xlabel('importancia')
plt.ylabel('Atributo o variables')
plt.show()
```
